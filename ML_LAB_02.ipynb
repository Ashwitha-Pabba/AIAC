 Identifying Numerical and Categorical Features


```python
numerical_features = []
categorical_features = []

# Iterate through columns and assign them based on dtype and context
for col in d.columns:
    if d[col].dtype == 'float64':
        numerical_features.append(col)
    elif d[col].dtype == 'int64':
        # For int64 columns, check if they are likely categorical (e.g., binary or few unique values)
        if d[col].nunique() <= 5 or col in ['anaemia', 'diabetes', 'high_blood_pressure', 'sex', 'smoking', 'DEATH_EVENT']:
            categorical_features.append(col)
        else:
            numerical_features.append(col)

print(f"Numerical Features: {numerical_features}")
print(f"Categorical Features: {categorical_features}")
```

## Interpret Distributions

### Analysis of 'Age' Distribution:

*   **Shape:** The distribution of 'Age' appears to be approximately normal, with a slight left skew. It is generally bell-shaped, with a peak around the mid-50s to early 60s.
*   **Central Tendency:** The majority of the patients fall within the age range of approximately 50 to 65. The peak of the distribution (mode) is around 55-60 years, which aligns with the calculated median age of 55.00.
*   **Spread:** The ages range from approximately 29 to 77. The standard deviation of 9.11 indicates a moderate spread around the mean.
*   **Outliers:** There don't appear to be significant outliers; the distribution smoothly tapers off at both ends, though there are fewer individuals at the youngest and oldest ends.

### Analysis of 'Cholesterol' Distribution:

*   **Shape:** The distribution of 'Cholesterol' is positively skewed (right-skewed). There is a sharp peak followed by a long tail to the right.
*   **Central Tendency:** The highest frequency of cholesterol values is observed around 200-250 mg/dL. The mode appears to be in this range. The calculated mean (not explicitly shown here but typically pulled towards the tail in a skewed distribution) would likely be higher than the median due to the right skew.
*   **Spread:** Cholesterol values range from 126 to 564. This indicates a wide spread, with a considerable difference between the minimum and maximum values.
*   **Outliers:** There are several data points that can be considered potential outliers or extreme values in the higher cholesterol ranges (e.g., above 400, particularly the value near 550), as they are far from the main cluster of data and appear in very small frequencies.
